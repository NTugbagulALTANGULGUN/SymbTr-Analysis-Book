# Contributing

## Overview

The **Turkish Delight Corpus (TDC) Analysis Book** is an open-source educational and research resource that supports reproducible computational analyses of Turkish makam music.

The project aims to promote open science by providing well-documented datasets, transparent computational workflows, and practical examples for statistical analysis, visualization, machine learning, and Music Information Retrieval (MIR).

Researchers, educators, students, and software developers are invited to contribute new analyses, educational materials, computational methods, documentation, and software improvements that enhance the quality, reproducibility, and long-term sustainability of the project.

## Ways to Contribute

Contributions to the **Turkish Delight Corpus (TDC) Analysis Book** are welcome from researchers, students, and practitioners interested in computational musicology, Music Information Retrieval (MIR), and symbolic music analysis.

Contributors are encouraged to:

- improve existing notebooks and computational workflows;
- develop new statistical analyses and visualization techniques;
- implement additional machine learning or artificial intelligence methods;
- integrate new open symbolic music datasets and benchmark resources;
- enhance documentation, tutorials, and educational content;
- identify and report issues, bugs, or inconsistencies;
- improve the reproducibility and efficiency of existing analyses;
- contribute new examples, case studies, or comparative experiments related to Turkish makam music.

## Coding Guidelines

To ensure consistency, reproducibility, and maintainability throughout the **Turkish Delight Corpus (TDC) Analysis Book**, contributors are encouraged to:

- write clear, modular, and well-documented Python code;
- follow the **PEP 8** style guidelines for Python programming;
- develop reproducible computational workflows;
- include explanatory Markdown cells describing each analysis step;
- generate tables and figures programmatically whenever possible;
- clearly document data preprocessing and transformation procedures;
- preserve the original dataset structure and avoid modifying raw data files;
- appropriately cite external datasets, software packages, and scientific publications;
- provide sufficient comments to facilitate understanding and future extension of the code.

## Reproducibility

Reproducibility is a fundamental principle of the Turkish Delight Corpus.

Contributors should ensure that notebooks:

- execute sequentially without manual intervention,
- use openly available datasets whenever possible,
- avoid hidden preprocessing steps,
- document software dependencies,
- generate all analyses programmatically.

## Citation

If you use the Turkish Delight Corpus in research or teaching, please cite the corresponding publication (when available) together with the original datasets and software resources employed throughout the analyses.

Proper citation supports reproducible research and acknowledges the valuable contributions of the original dataset developers.

## Acknowledgements

The Turkish Delight Corpus builds upon the outstanding datasets, software tools, and computational musicology research developed by the Music Technology Group (MTG), Universitat Pompeu Fabra (UPF), together with contributions from the broader Music Information Retrieval community.

Their commitment to open science has made many of the analyses presented throughout this book possible.
