# Literature Review

## Introduction

Turkish makam music represents a historically rich modal tradition characterized by microtonal pitch organization, makam-specific melodic development (*seyir*), characteristic pitches, modulation practices, and cyclic rhythmic structures known as *usul*. These musical characteristics distinguish Turkish makam music from Western tonal systems and create specific challenges for computational representation, analysis, retrieval, and modeling {cite}`bozkurt2014review`.

The emergence of Music Information Retrieval (MIR) and computational musicology has enabled musical collections to be studied using machine-readable representations, statistical methods, signal-processing techniques, and data-driven models. MIR encompasses the computational organization, retrieval, description, comparison, and analysis of musical information represented through symbolic scores, audio recordings, metadata, or combinations of these modalities {cite}`downie2003mir,casey2008cbmir,muller2015fmp`.

Nevertheless, computational methods developed primarily for Western music cannot always be transferred directly to culturally distinct musical traditions. Musical concepts, tuning systems, rhythmic organizations, notation conventions, and performance practices vary considerably across cultures. Therefore, culture-specific computational models and representative research corpora are required to analyze Turkish makam music without reducing its musical structure to assumptions derived exclusively from Western tonal theory {cite}`serra2011multicultural,serra2014creating`.

A major development in this field was the construction of publicly accessible corpora containing Turkish makam music scores and recordings. The introduction of the SymbTr symbolic database provided researchers with a machine-readable collection designed specifically for computational analysis and MIR applications {cite}`karaosmanoglu2012symbtr`. Related corpus-development initiatives subsequently expanded the available research material and supported studies addressing tonic identification, makam recognition, melodic segmentation, structural analysis, audio–score alignment, and performance analysis {cite}`uyar2014corpus,karaosmanoglu2014phrases,senturk2013tonic,senturk2014linking`.

More recent developments in machine learning, deep learning, transformer architectures, and foundation models have created additional opportunities for learning musical representations, identifying structural patterns, and generating symbolic music. However, their application to Turkish makam music requires sufficiently large datasets, culturally informed representations, transparent evaluation procedures, and models capable of preserving makam-specific melodic and microtonal characteristics {cite}`goodfellow2016dl,lecun2015deep,vaswani2017attention,bommasani2021foundation,huang2025musicfm`.

This chapter reviews the literature according to the historical and methodological development of computational Turkish makam music research. It begins with the historical foundations of makam theory and the transition from traditional performance to symbolic representation. It then examines symbolic corpora, makam and tonic analysis, melodic segmentation, audio–score alignment, Music Information Retrieval, statistical learning, deep learning, and emerging foundation-model approaches. The chapter concludes by identifying persistent limitations and research opportunities relevant to future computational studies of Turkish makam music.

## Historical Foundations of Turkish Makam Theory

The theoretical foundations of Turkish makam music developed over several centuries through the work of scholars, theorists, and musicians who sought to explain modal organization, melodic progression, intervallic relationships, and rhythmic structures. These early theoretical studies gradually shaped the conceptual framework of the makam system and established many of the musical principles that continue to characterize Turkish classical music today.

Unlike Western tonal music, Turkish makam music is primarily organized around melodic development rather than harmonic progression. Concepts such as *makam*, *seyir*, characteristic pitches, modulation, and *usul* constitute the core elements of the tradition and later became essential concepts in computational analysis. Understanding these musical characteristics is therefore fundamental for developing digital representations and computational models of Turkish makam music.

The following sections briefly summarize the historical evolution of makam theory before introducing its transition into computational musicology.

### Al-Fārābī and the Early Systematization of Music Theory

Among the earliest scholars associated with the scientific study of music, Al-Fārābī occupies a prominent place in the history of music theory. His writings treated music as a scientific discipline and examined topics including intervals, scales, rhythm, and the mathematical relationships underlying musical sound. His theoretical approach influenced later generations of music scholars and contributed to the development of modal thinking throughout the Islamic world.

Although the modern Turkish makam system had not yet emerged, these early theoretical ideas established an intellectual foundation upon which later modal theories were developed. Consequently, Al-Fārābī is widely recognized as one of the important historical figures in the evolution of music theory.

### Ṣafī al-Dīn al-Urmawī and *Kitāb al-Adwār*

Several centuries later, Ṣafī al-Dīn al-Urmawī further advanced modal music theory by presenting a more systematic description of scales, intervals, rhythmic cycles, and modal organization. His theoretical framework influenced musical thought across a wide geographical region and served as an important reference for later developments in Ottoman music theory.

Many concepts that later became associated with the Turkish makam tradition evolved from this theoretical heritage. Although contemporary makam theory has undergone significant development, the systematic treatment of modal organization introduced during this period represents an important stage in the historical evolution of Turkish music theory.

### From Historical Music Theory to Computational Musicology

For centuries, the transmission of Turkish makam music relied primarily on oral tradition, master–apprentice relationships, theoretical treatises, and written notation. While these resources have preserved an extensive musical heritage, they also present challenges for large-scale analysis, comparison, and quantitative investigation. As the number of available musical scores and recordings increased, computational approaches emerged as an effective means of studying the structural characteristics of makam music.

The development of digital music representations enabled musical compositions to be stored in machine-readable formats and analyzed using computational methods. Instead of examining individual compositions manually, researchers became able to investigate thousands of musical pieces simultaneously, facilitating statistical analysis, pattern discovery, similarity measurement, and automatic music processing.

This transition marked the beginning of computational musicology for Turkish makam music. Research gradually expanded from the digitization of musical scores to the construction of symbolic corpora, automatic makam recognition, melodic analysis, phrase segmentation, audio–score alignment, and machine learning techniques. These developments established the methodological foundation of contemporary computational research on Turkish makam music and continue to support recent advances in artificial intelligence and generative music systems.

## Organization of the Literature

The computational study of Turkish makam music has developed through several interconnected research directions. Early studies primarily focused on the digitization of musical scores and the construction of machine-readable symbolic corpora. These resources subsequently enabled systematic investigations of makam structure, tonic identification, pitch organization, melodic progression, phrase boundaries, and musical similarity.

As digital collections expanded, research moved beyond symbolic representation toward multimodal analysis involving both musical scores and audio recordings. Audio–score alignment methods supported the synchronization of notated compositions with recorded performances and made it possible to investigate timing, tuning, ornamentation, and performer-specific interpretation.

Music Information Retrieval provided a broader methodological framework for organizing, comparing, retrieving, and classifying makam music. Statistical techniques and classical machine learning methods were then employed for feature-based analysis, makam classification, clustering, and pattern discovery. More recently, deep learning and transformer-based architectures have introduced data-driven representation learning and generative modeling approaches.

Accordingly, the literature reviewed in this chapter is organized into eight principal research areas:

1. digitalization and symbolic music corpora,
2. makam, tonic, and melodic analysis,
3. phrase segmentation and structural analysis,
4. audio–score alignment,
5. Music Information Retrieval,
6. statistical and machine learning approaches,
7. deep learning, and
8. foundation models and generative artificial intelligence.

This thematic organization reflects both the historical evolution of the field and the increasing methodological complexity of computational research on Turkish makam music.

## 1. Digitalization and Symbolic Music Corpora

The development of computational methods for Turkish makam music has been closely linked to the availability of machine-readable symbolic datasets. Unlike audio recordings, symbolic representations explicitly encode musical events such as pitches, note durations, measures, rhythmic structures, and metadata, allowing researchers to investigate musical characteristics algorithmically. Consequently, the creation of symbolic corpora has become the foundation of almost every subsequent study in computational musicology, including automatic makam recognition, melodic analysis, phrase segmentation, similarity measurement, and machine learning.

One of the earliest milestones in this direction was the introduction of **SymbTr**, the first comprehensive symbolic database specifically designed for Turkish makam music research. The dataset transformed traditional musical notation into a standardized machine-readable representation suitable for computational analysis and Music Information Retrieval (MIR). By providing consistent symbolic encoding together with rich musical metadata, SymbTr enabled researchers to investigate large collections of compositions using reproducible computational methods rather than manual score inspection {cite}`karaosmanoglu2012symbtr`.

The construction of symbolic datasets was further supported through the CompMusic project, whose objective was to establish research infrastructures for culturally diverse musical traditions. Instead of adapting computational techniques developed exclusively for Western music, the project emphasized the creation of representative corpora that preserve the theoretical and musical characteristics of non-Western traditions. Within this framework, Turkish makam music became one of the principal case studies, demonstrating how carefully curated symbolic datasets can support reproducible computational musicology and cross-cultural MIR research {cite}`serra2014creating`.

As the availability of symbolic collections increased, researchers recognized that dataset quality depends not only on the number of musical pieces but also on the consistency of symbolic encoding, metadata organization, annotation quality, and data validation procedures. Consequently, additional corpus development studies focused on improving data standardization and expanding the available resources for computational analysis. These efforts significantly improved the usability of Turkish makam music datasets for statistical analysis, machine learning, and information retrieval applications {cite}`uyar2014corpus`.

Another important contribution was the introduction of phrase-level symbolic datasets. Rather than representing complete musical compositions only, these datasets provided manually annotated melodic phrases, enabling researchers to investigate structural organization at finer temporal resolutions. Phrase annotations subsequently became an important resource for melodic segmentation, motif discovery, structural analysis, and similarity measurement, extending the scope of symbolic music research beyond complete-score analysis {cite}`karaosmanoglu2014phrases`.

The continuous evolution of symbolic resources has recently culminated in **SymbTr v3.0**, which substantially expands previous releases through a larger collection of Turkish makam compositions and multiple interoperable symbolic formats. The current version includes MusicXML, MIDI, PDF, TXT, Mu2, and related metadata, providing a comprehensive research resource suitable for reproducible computational studies, educational purposes, machine learning, and generative artificial intelligence research. The availability of openly accessible, standardized symbolic corpora considerably lowers the barrier for future studies and facilitates collaboration across the international Music Information Retrieval community {cite}`karaosmanoglu2025symbtrv3`.

Overall, the progressive development of symbolic corpora established the data infrastructure upon which nearly all subsequent computational studies of Turkish makam music have been built. Later research on makam recognition, tonic identification, audio–score alignment, structural analysis, and deep learning directly benefited from these increasingly comprehensive symbolic resources, illustrating the central role of dataset construction in the advancement of computational musicology.

An important advancement in symbolic corpus construction was the introduction of phrase-level annotation. Instead of representing only complete musical scores, researchers manually identified melodic phrase boundaries and annotated phrase structures according to musicological knowledge. These annotations enabled computational studies of melodic segmentation, motif discovery, structural similarity, and phrase-level retrieval. The resulting phrase dataset has since become an important benchmark for structural analysis in Turkish makam music and demonstrates how expert musicological annotations can be integrated into symbolic corpora for computational research {cite}`karaosmanoglu2014phrases`.


The studies summarized in Table {numref}`tab-symbolic-corpora` illustrate the evolution of symbolic corpus development for Turkish makam music. Starting from the introduction of standardized symbolic representations, subsequent research expanded these resources through metadata enrichment, phrase annotations, structural information, and interoperable formats. Together, these studies established the computational foundation for reproducible Music Information Retrieval and artificial intelligence research.

:::{table} Representative studies on symbolic corpora for Turkish makam music.
:name: tab-symbolic-corpora

| Year | Reference | Main Contribution | Research Area |
|:---:|-----------|------------------|---------------|
|2012|{cite}`karaosmanoglu2012symbtr`|Introduced the SymbTr symbolic corpus.|Corpus construction|
|2014|{cite}`uyar2014corpus`|Extended corpus organization and metadata.|Dataset development|
|2014|{cite}`karaosmanoglu2014phrases`|Published phrase-level annotations.|Structural annotation|
|2025|{cite}`karaosmanoglu2025symbtrv3`|Released SymbTr v3.0 with interoperable formats.|Open benchmark dataset|

:::

## 2. Makam, Tonic, and Melodic Analysis

The availability of large symbolic corpora significantly accelerated research on the automatic analysis of Turkish makam music. Rather than focusing solely on corpus construction, researchers began developing computational methods capable of identifying makam, estimating the tonic, characterizing melodic progression (*seyir*), and analyzing intervallic relationships. These studies aimed to model the theoretical concepts of Turkish makam music using reproducible computational techniques while preserving the distinctive characteristics of the musical tradition.

One of the earliest research directions concentrated on automatic makam recognition. Unlike genre or instrument classification, makam recognition requires algorithms to capture subtle melodic characteristics, intervallic relationships, and pitch organizations that distinguish one makam from another. Researchers therefore investigated statistical pitch representations, melodic descriptors, and culture-specific computational features capable of modeling makam identity more effectively than conventional Western music descriptors {cite}`senturk2013tonic,bozkurt2014review`.

Tonic identification soon emerged as another fundamental research problem because many computational analyses depend on correctly determining the tonal center of a musical composition. Since pitch distributions vary considerably across performances and transpositions, automatic tonic estimation became an essential preprocessing step for makam recognition, melodic comparison, and large-scale corpus analysis. Data-driven approaches demonstrated that combining pitch-distribution analysis with musicological knowledge substantially improves the robustness of automatic tonic estimation {cite}`senturk2013tonic`.

As research progressed, attention gradually shifted from static pitch distributions toward the dynamic behavior of melodies. In Turkish makam music, melodic identity is not determined solely by the collection of pitches but also by the characteristic progression of melodic phrases known as *seyir*. Consequently, computational studies increasingly incorporated sequential information, melodic contour analysis, and structural relationships to model makam-specific melodic development more accurately. These investigations demonstrated that melodic progression provides complementary information beyond conventional statistical pitch representations and therefore contributes to more reliable computational analysis {cite}`bozkurt2014review`.

The availability of phrase-level symbolic datasets further strengthened melodic analysis by providing manually annotated structural units. Instead of analyzing complete compositions only, researchers became able to investigate phrase organization, melodic similarity, motif repetition, and structural variation at finer temporal resolutions. Such annotations not only facilitated musicological analysis but also created valuable benchmark datasets for evaluating automatic segmentation algorithms and machine learning models {cite}`karaosmanoglu2014phrases`.

Collectively, these studies established the methodological foundation for computational makam analysis. They demonstrated that successful modeling of Turkish makam music requires the integration of statistical analysis, symbolic representations, musicological knowledge, and computational learning techniques. These methodological advances subsequently supported more sophisticated research in Music Information Retrieval, audio–score alignment, deep learning, and generative artificial intelligence.

Table {numref}`tab-makam-analysis` summarizes representative studies on automatic makam recognition, tonic estimation, melodic analysis, and computational music theory. These works collectively established the foundations for symbolic analysis of Turkish makam music.

:::{table} Representative studies on makam analysis.
:name: tab-makam-analysis

| Year | Reference | Main Contribution | Task |
|:---:|-----------|------------------|------|
|2013|{cite}`senturk2013tonic`|Automatic tonic identification.|Tonic estimation|
|2014|{cite}`atli2014features`|Pitch feature extraction.|Feature engineering|
|2015|{cite}`atli2015tonic`|Improved tonic estimation.|Tonic detection|
|2014|{cite}`Bozkurt2014JNMR`|Computational makam analysis.|Makam recognition|
|2014|{cite}`Bozkurt2014SIU`|Pitch distribution analysis.|Feature analysis|
|2014|{cite}`Bozkurt2014AAWM`|Computational makam modeling.|Music theory|

:::

## 3. Phrase Segmentation and Structural Analysis

Beyond identifying makam and tonic, researchers increasingly focused on understanding the internal structure of Turkish makam compositions. Musical pieces are naturally organized into phrases that represent coherent melodic units and play a fundamental role in performance, analysis, and music education. Automatically identifying these structural boundaries enables computational systems to analyze musical organization at multiple temporal scales and supports more detailed investigations of melodic development.

Early computational studies relied primarily on expert musicological annotations to establish reliable phrase boundaries. These manually annotated datasets provided high-quality reference material for evaluating automatic segmentation algorithms and created an essential bridge between traditional musicological expertise and computational analysis. By preserving musicological knowledge within machine-readable symbolic corpora, phrase annotations enabled reproducible experiments and comparative evaluation across different computational methods {cite}`karaosmanoglu2014phrases`.

Building upon these annotated resources, subsequent research explored automatic melodic segmentation techniques based on symbolic note sequences, interval relationships, rhythmic organization, and statistical properties of melodic progression. Rather than considering individual notes independently, these approaches attempted to identify musically meaningful structural units that correspond to human perception of melodic phrases. Such segmentation methods substantially reduced the amount of manual annotation required while providing scalable solutions for large symbolic collections.

Structural analysis further expanded beyond phrase boundary detection to include motif discovery, melodic similarity, and repeated-pattern analysis. Since Turkish makam music frequently contains characteristic melodic formulas that recur across different compositions, identifying these recurring structures became an important research objective. Computational comparison of melodic fragments enables researchers to investigate stylistic relationships among composers, examine makam-specific melodic behavior, and explore the evolution of musical traditions through large symbolic corpora.

Phrase-level representations also proved valuable for downstream machine learning applications. Instead of learning directly from complete musical compositions, algorithms can be trained on shorter and structurally meaningful melodic segments. This approach often improves representation learning, reduces computational complexity, and provides models with musically interpretable units that better reflect the hierarchical organization of Turkish makam music.

Overall, phrase segmentation and structural analysis represent an important intermediate layer between low-level symbolic notation and higher-level musical understanding. Together with symbolic corpora, these studies provide the structural foundation required for similarity analysis, Music Information Retrieval, sequence modeling, and modern artificial intelligence approaches developed for Turkish makam music.

Table {numref}`tab-phrase-analysis` summarizes representative studies addressing phrase segmentation and structural analysis. These studies introduced annotated phrase datasets and computational methods for identifying phrase boundaries and modeling musical structure.

:::{table} Representative studies on phrase segmentation.
:name: tab-phrase-analysis

| Year | Reference | Main Contribution | Task |
|:---:|-----------|------------------|------|
|2014|{cite}`karaosmanoglu2014phrases`|Phrase-level annotations.|Dataset|
|2015|{cite}`holzapfel2015section`|Automatic section detection.|Segmentation|
|2016|{cite}`senturk2016structure`|Structural analysis.|Structure modeling|
|2016|{cite}`senturk2016thesis`|Comprehensive computational framework.|PhD thesis|

:::

## 4. Audio–Score Alignment

While symbolic music corpora provide explicit descriptions of musical structure, they cannot fully capture the expressive characteristics of live performance. Turkish makam performances exhibit variations in timing, ornamentation, intonation, tempo, articulation, and improvisation that are not completely represented in symbolic notation. Consequently, researchers have increasingly investigated methods capable of synchronizing symbolic scores with audio recordings, thereby establishing direct correspondences between written notation and performed music.

Audio–score alignment aims to determine the temporal relationship between symbolic musical events and their corresponding positions within an audio recording. Accurate synchronization enables detailed investigation of expressive performance characteristics while preserving the structural information available in symbolic notation. Compared with Western classical music, alignment in Turkish makam music presents additional challenges due to microtonal intervals, flexible rhythmic interpretation, ornamentation, and performer-dependent timing variations. These characteristics require alignment algorithms that incorporate both computational techniques and musicological knowledge {cite}`senturk2014linking`.

Early studies demonstrated that combining score-informed representations with automatic alignment algorithms substantially improves synchronization accuracy for Turkish makam music. Rather than relying exclusively on generic signal-processing techniques, these approaches exploited symbolic information extracted from musical scores to guide audio analysis. The resulting systems established reliable note-level correspondences between symbolic notation and performed audio, creating valuable resources for computational musicology and Music Information Retrieval applications {cite}`senturk2014linking`.

The availability of aligned symbolic and audio data subsequently enabled a broad range of performance analysis studies. Researchers became able to investigate expressive timing, pitch deviations, ornamentation patterns, tempo fluctuations, and performer-specific interpretation using objective computational methods. These analyses provide important insights into the relationship between theoretical notation and actual musical practice while supporting comparative studies across performers, recordings, and historical interpretations.

Audio–score alignment has also become an important enabling technology for downstream computational tasks. Accurate synchronization facilitates automatic transcription, melodic similarity analysis, performer comparison, music education systems, interactive visualization, and multimodal machine learning models that jointly exploit symbolic notation and audio recordings. Furthermore, aligned datasets provide valuable training data for deep learning methods seeking to learn robust representations across multiple musical modalities.

Overall, audio–score alignment bridges the gap between symbolic music representations and real-world musical performance. By integrating notation with expressive audio recordings, these methods considerably expand the scope of computational analysis and provide an essential foundation for multimodal Music Information Retrieval and artificial intelligence research on Turkish makam music.
To perform reliable alignment, symbolic scores and audio recordings must first be transformed into comparable computational representations. Rather than comparing notation and waveform directly, alignment algorithms extract descriptive features from both modalities that emphasize musically meaningful information while reducing irrelevant variability. Figure {numref}`fig-score-audio-features` illustrates the principal symbolic and audio representations employed in the alignment framework, together with the corresponding feature extraction process {cite}`senturk2014linking`.

```{figure} images/score_audio_features.png
---
width: 100%
align: center
name: fig-score-audio-features
---
Comparison of symbolic score and audio representations used for score–audio alignment in Turkish makam music. The figure shows the symbolic score, the corresponding audio recording, the prominent-pitch representations derived from both modalities, and the Harmonic Pitch Class Profile (HPCP) features employed to establish robust synchronization between symbolic notation and performed audio. Adapted from Şentürk et al. {cite}`senturk2014linking`.
```

## 5. Music Information Retrieval

The rapid growth of symbolic corpora and aligned audio–score resources considerably expanded the scope of Music Information Retrieval (MIR) research for Turkish makam music. Rather than focusing solely on the storage and digitization of musical materials, researchers increasingly developed computational methods capable of organizing, searching, comparing, and analyzing large music collections. These developments positioned Turkish makam music within the broader MIR community while simultaneously highlighting the need for computational approaches that respect culture-specific musical characteristics.

One of the most influential initiatives in this direction was the **CompMusic** project, which aimed to establish computational methodologies for several non-Western musical traditions. Instead of adapting algorithms originally developed for Western tonal music, CompMusic emphasized the importance of culturally informed representations that preserve the theoretical and performance characteristics of traditions such as Turkish makam music, Hindustani music, Carnatic music, and Arab-Andalusian music. The project demonstrated that successful MIR systems require not only efficient algorithms but also representative datasets, appropriate musicological knowledge, and standardized evaluation protocols {cite}`serra2014creating`.

Within this framework, symbolic corpora, audio recordings, metadata, and music-theoretical information were integrated into unified computational environments. These resources enabled researchers to investigate melodic similarity, makam classification, tonic identification, score-informed retrieval, performance analysis, and multimodal music exploration. The integration of multiple information sources significantly improved the capability of MIR systems to capture the complex musical structures that characterize Turkish makam music {cite}`senturk2014linking,karaosmanoglu2025symbtrv3`.

Another important contribution of MIR research has been the development of computational similarity measures specifically designed for Turkish makam music. Unlike Western tonal repertoires, similarity within makam music depends not only on shared pitch content but also on melodic progression (*seyir*), phrase organization, ornamentation, rhythmic structure, and modal development. Consequently, researchers increasingly incorporated symbolic representations, sequential analysis, and musicological knowledge into similarity estimation methods, producing retrieval systems that more accurately reflect human musical perception.

The growing availability of interoperable symbolic datasets has also facilitated reproducible benchmarking within the MIR community. Openly accessible corpora allow researchers to evaluate algorithms under comparable experimental conditions, encouraging transparent comparison between competing methods and accelerating methodological development. This emphasis on reproducibility has become particularly important as computational approaches continue to evolve toward increasingly complex machine learning and deep learning models {cite}`karaosmanoglu2025symbtrv3`.

Overall, MIR research has transformed Turkish makam music from a collection of individually studied compositions into a searchable, analyzable, and computationally accessible musical knowledge base. These developments established the methodological infrastructure upon which modern statistical learning, deep neural networks, and foundation models are now being developed.

## 6. Statistical and Machine Learning Approaches

The increasing availability of symbolic datasets enabled the application of statistical analysis and machine learning techniques to Turkish makam music. Early studies primarily relied on handcrafted musical features derived from symbolic scores and musicological knowledge. These features were then used to investigate makam classification, tonic identification, melodic similarity, and structural characteristics through supervised and unsupervised learning approaches {cite}`bozkurt2014review`.

Feature engineering played a central role in these studies. Researchers extracted pitch distributions, interval statistics, melodic contours, rhythmic patterns, and phrase-level descriptors to represent musical pieces numerically. These representations provided interpretable features that reflected both theoretical knowledge and musical practice {cite}`bozkurt2014review,karaosmanoglu2014phrases`.

Classical machine learning algorithms, including Support Vector Machines (SVM), k-Nearest Neighbors (k-NN), Decision Trees, and Random Forests, were subsequently employed for automatic makam classification and related prediction tasks. These methods demonstrated that symbolic musical features could effectively discriminate between different makams when combined with appropriate feature selection strategies {cite}`bozkurt2014review`.

Unsupervised learning methods also became increasingly important for exploring the internal organization of Turkish makam corpora. Clustering and dimensionality reduction techniques such as Principal Component Analysis (PCA) enabled researchers to identify hidden musical structures, visualize relationships among compositions, and investigate stylistic similarities across large collections without predefined class labels.

Overall, statistical learning established the methodological bridge between traditional rule-based computational musicology and modern deep learning approaches. The experience gained from feature engineering and classical machine learning later informed the development of neural network architectures capable of learning musical representations directly from symbolic data.

## 7. Deep Learning

Recent advances in deep learning have significantly influenced computational research on Turkish makam music. Unlike classical machine learning methods that depend on handcrafted musical features, deep neural networks learn hierarchical representations directly from symbolic sequences or audio data. This capability enables models to capture complex melodic and structural relationships with minimal manual feature engineering {cite}`goodfellow2016dl,lecun2015deep`.

Sequential neural architectures, including recurrent neural networks (RNNs), Long Short-Term Memory (LSTM) networks, and Transformer models, have demonstrated strong potential for modeling melodic progression and long-term musical dependencies. These models are particularly suitable for Turkish makam music because they learn sequential characteristics directly from symbolic note sequences.

Deep learning has also expanded the scope of symbolic music generation, representation learning, and multimodal analysis. However, the effectiveness of these models remains strongly dependent on the availability of sufficiently large and well-annotated symbolic corpora, highlighting the continued importance of open datasets such as SymbTr {cite}`karaosmanoglu2025symbtrv3`.

## 8. Foundation Models and Generative Artificial Intelligence

Foundation models have recently emerged as one of the most influential paradigms in artificial intelligence, enabling models to learn transferable representations from large-scale datasets through self-supervised pretraining. Unlike conventional deep learning models that are typically designed for a single task, foundation models can be adapted to a wide range of downstream applications, including classification, retrieval, generation, and multimodal understanding. In computational musicology, this paradigm has enabled the development of models capable of jointly processing symbolic music, audio recordings, textual descriptions, and metadata, thereby supporting more comprehensive music understanding and analysis {cite}`bommasani2021foundation`.

Building upon these advances, recent music-specific foundation models have demonstrated the ability to learn general-purpose musical representations from diverse symbolic and audio corpora. These models support a variety of downstream tasks, including music classification, similarity retrieval, representation learning, symbolic music generation, and multimodal music analysis. Compared with conventional deep learning approaches, foundation models offer improved transferability and scalability by leveraging knowledge acquired during large-scale pretraining {cite}`huang2025musicfm`.

In the context of Turkish makam music, the application of foundation models remains at an early stage, primarily due to the limited availability of large-scale annotated symbolic datasets. Nevertheless, the continuous development of open resources such as SymbTr v3.0, together with richer metadata, phrase-level annotations, and interoperable symbolic representations, provides a promising foundation for future research. These resources are expected to facilitate the development of culturally informed foundation models capable of preserving the unique melodic, modal, and structural characteristics of Turkish makam music while supporting advanced retrieval, analysis, and generative music applications {cite}`karaosmanoglu2025symbtrv3`.

## 9. Summary and Research Gaps

Computational research on Turkish makam music has evolved from symbolic corpus construction to advanced machine learning and artificial intelligence methods. The development of datasets such as SymbTr, phrase-level annotations, and score–audio alignment frameworks has established a reproducible computational infrastructure supporting Music Information Retrieval, statistical analysis, deep learning, and generative AI research.

Despite this progress, several challenges remain. Existing datasets are still relatively small compared with those available for Western music, limiting the effective training of large neural networks and foundation models. In addition, the scarcity of standardized benchmarks, multimodal datasets, and comprehensive evaluation protocols makes direct comparison between computational methods difficult. Future work should therefore prioritize larger open datasets, multimodal corpus construction, explainable AI, reproducible evaluation frameworks, and culturally informed foundation models capable of accurately representing the unique characteristics of Turkish makam music {cite}`karaosmanoglu2025symbtrv3,bommasani2021foundation,huang2025musicfm`.
