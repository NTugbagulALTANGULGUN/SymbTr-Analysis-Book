# Open Research Problems

## Overview

Despite significant progress in computational analysis of Turkish makam music, several important research challenges remain open.

This chapter summarizes representative open problems that may guide future studies in computational musicology, Music Information Retrieval (MIR), symbolic music processing, and artificial intelligence.

## Remaining Challenges

### Limited Benchmark Datasets

Although several symbolic datasets have been published for Turkish makam music, only a limited number have become widely adopted as benchmark resources. Existing corpora differ considerably in repertoire coverage, annotation quality, symbolic representation, metadata completeness, and licensing policies, reducing reproducibility and making objective comparison between computational methods more difficult.

### Limited Multimodal Resources

Current research is primarily based on symbolic scores or audio recordings considered independently. Comparatively few datasets integrate symbolic notation, audio performances, lyrics, metadata, performer information, historical context, and expert musicological annotations within a unified framework. Such multimodal resources would facilitate richer computational analyses and support the development of more comprehensive artificial intelligence models.

### Scarcity of Large-Scale Pretrained and Foundation Models

While large pretrained models have transformed natural language processing, computer vision, and Western Music Information Retrieval, equivalent resources remain scarce for Turkish makam music and related modal traditions. The absence of culturally informed pretrained models limits the adoption of transfer learning, self-supervised learning, multimodal representation learning, and foundation models in this domain.

### Explainable and Interpretable Artificial Intelligence

Many contemporary machine-learning models achieve high predictive performance but provide limited insight into their decision-making processes. Future computational systems should explain which melodic phrases, pitch organizations, rhythmic structures, cadential patterns, or modal characteristics contribute to makam recognition and classification, thereby strengthening the connection between computational analysis and established music theory.

### Reproducible Evaluation Protocols

Current studies frequently employ different preprocessing strategies, train–test partitioning methods, feature extraction procedures, and evaluation metrics. In some cases, data leakage may occur when highly similar musical excerpts appear in both training and testing sets. Standardized, transparent, and leakage-free evaluation protocols are therefore essential for ensuring fair comparison between computational approaches.

### Class Imbalance and Repertoire Bias

Available corpora often exhibit highly uneven distributions across makam categories, composers, musical forms, historical periods, and performance traditions. These imbalances may bias machine-learning models and reduce their ability to generalize to underrepresented repertoires.

### Metadata Quality and Standardization

Differences in makam nomenclature, transposition practices, compound makam definitions, theoretical classifications, and metadata completeness continue to present significant challenges for supervised learning and large-scale computational analysis. Improved metadata standards would substantially enhance dataset interoperability and reproducibility.

### Integration of Computational and Musicological Knowledge

Future research should increasingly combine data-driven artificial intelligence techniques with expert musicological knowledge. Incorporating concepts such as *seyir*, *çeşni*, *karar*, *güçlü*, modulation patterns, and performance practice into computational models will produce systems that are both more accurate and more interpretable.

## Discussion

The open research problems presented in this chapter demonstrate that computational analysis of Turkish makam music is entering a new stage of development. Although significant progress has been achieved through the creation of symbolic corpora, phrase-level datasets, and machine learning approaches, several important challenges remain to be addressed.

Future research should focus on the development of larger and more standardized benchmark datasets, multimodal learning frameworks that integrate symbolic scores with audio recordings, and culturally informed foundation models capable of learning meaningful representations of Turkish makam music. In addition, explainable artificial intelligence, standardized evaluation protocols, and open, reproducible computational workflows will be essential for improving the reliability, transparency, and comparability of future studies.

The **Turkish Delight Corpus (TDC) Analysis Book** contributes to this research direction by providing a reproducible computational environment for exploring symbolic Turkish makam music datasets. By combining dataset exploration, structural analysis, statistical evaluation, visualization, and practical implementation examples, it establishes a foundation that can support future research in computational musicology, Music Information Retrieval (MIR), and artificial intelligence for culturally diverse musical traditions.

## Future Outlook

Future progress in computational Turkish makam music research will depend on the development of larger, more diverse, and carefully curated benchmark datasets integrating symbolic notation, audio recordings, lyrics, metadata, historical information, and expert musicological annotations. Such resources will provide the foundation for reproducible evaluation and facilitate the comparison of emerging computational methods.

Advances in self-supervised learning, transfer learning, multimodal representation learning, and foundation models offer promising opportunities for reducing dependence on manually labeled datasets while enabling richer representations of modal music traditions. These approaches have the potential to improve makam recognition, melodic similarity analysis, phrase segmentation, symbolic music generation, transcription, retrieval, and interactive music analysis.

Another important research direction concerns the development of explainable and culturally informed artificial intelligence. Rather than predicting makam labels alone, future systems should identify the musical evidence underlying their decisions and present computational results in ways that are meaningful to both musicologists and performers.

Graph neural networks, transformer architectures, long-context sequence models, and multimodal learning frameworks are expected to play an increasingly important role in modeling long-term melodic relationships and complex modal structures. These technologies may also contribute to intelligent educational systems, computational composition, digital cultural heritage preservation, and interactive music discovery.

Finally, the long-term success of the field will depend on transparent and reproducible research practices. Publicly available datasets, open-source software, documented preprocessing pipelines, standardized evaluation protocols, version-controlled experiments, and openly accessible benchmark results will be essential for establishing Turkish makam music as a mature and sustainable research domain within computational musicology and artificial intelligence.

## Summary

The open problems presented in this chapter highlight that computational research on Turkish makam music continues to offer significant opportunities for methodological and technological advancement.

Addressing these challenges will require interdisciplinary collaboration among musicologists, musicians, computer scientists, machine learning researchers, and digital humanities scholars. Progress in areas such as benchmark dataset development, symbolic music processing, explainable artificial intelligence, multimodal learning, and reproducible research will further strengthen the field.

The **Turkish Delight Corpus (TDC) Analysis Book** contributes to this effort by providing an open, reproducible, and extensible computational framework for exploring symbolic Turkish makam music datasets. By combining dataset exploration, statistical analysis, visualization, and practical code examples, it aims to support future research, education, and the development of advanced Music Information Retrieval (MIR) and machine learning methodologies.

## Key Takeaways

> **Key points**
>
> - Standardized benchmark datasets are essential for reproducible and comparable computational research on Turkish makam music.
> - Symbolic music corpora provide a strong foundation for statistical analysis, visualization, and machine learning applications.
> - Future research should emphasize multimodal learning, culturally informed foundation models, and explainable artificial intelligence.
> - Open datasets, reproducible computational workflows, and collaborative research practices can accelerate advances in computational musicology and Music Information Retrieval (MIR).
